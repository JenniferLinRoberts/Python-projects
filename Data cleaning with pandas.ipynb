## The goal is to clean this data set so that it can be given to call centre employes.
### First dropping any duplucates

```python
df = df.drop_duplicates()
```

### Now removing any unneeded columns.

```python
df = df.drop(columns = 'Not_Useful_Column')
```

### Now going column by column to see what needs fixing. Staring from the left. 
#### Customer ID and First_name look fine. Last_Name has some problems. Using str.strip to remove the punctuation from the last names. (note: the strip function is from regular expression which can get complex so here each punctuation is taken out one by one using the left or right string strip)
#### note: it is important that only the column being worked on is updated. In this case the Last_Name column is updated. if the whole data frame is updated, all other columns would be erased as the function being used is only working on the Last_Name column. 

```python
df['Last_Name'] = df['Last_Name'].str.lstrip('...')
df['Last_Name'] = df['Last_Name'].str.lstrip('/')
df['Last_Name'] = df['Last_Name'].str.rstrip('_')
```

#### To run all of this in one line you can use the following 

```python
# df['Last_Name'] = df['Last_Name'].str.strip('123._/')
```

#### Moving on to the Phone_Number column. 
#### First non numeric values will be removed then the phone numbers will be formatted in a standardised way.

```python
# anything that is not alphabetic values upper or lower and numbers 0-9 will be replaced with nothing.
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '')  
```

#### Now the Phone numbers are just plane numbers the chosen formatting can be added in.
#### Adding in the formatting. 
#### First the values in the Phone Number column need to be changed to strings (they are currently numbers). To do this, the string coversion needs to applyed to each individual value. Applying the string convertion to the entire column wont work (this wont work: str(df['Phone_Number)']). 

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))  #This could be done in a for loop but lambda is quicker.
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3]+'-' + x[3:6]+'-' + x[6:10] )
```

#### Now the NA values can be removed 

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('nan--', '')
df['Phone_Number'] = df['Phone_Number'].str.replace('Na--', '')
```

#### Moving on to the Adress column. 
#### The adress column will be split so that the zip code has its own column to make it more readable for the call centre staff

```python
df[['Street_Address', 'State', 'Zip_Code']]=df['Address'].str.split(',', 2, expand = True)
```

#### Now the formatting for the Paying Customer column will be standardised. Same for the Do Not Contact column.

```python
df['Paying Customer'] = df['Paying Customer'].str.replace('Yes', 'Y')
df['Paying Customer'] = df['Paying Customer'].str.replace('No', 'N')

df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No', 'N')
```

### Now all the individual columns are fixed, the data set as a whole can be considered. 
### Now all the NA and NaN values can be standardised

```python
df = df.replace('N/A','' )
df = df.replace('N/a','' )
df = df.fillna('') #the NaN values are not actually values, they are nulls so must be treated differently
```
