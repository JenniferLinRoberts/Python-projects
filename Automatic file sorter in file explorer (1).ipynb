```python
import os, shutil
```

```python
#the r before the string is needed because of the back slashes 
path = r"C:/Users/Admin/OneDrive/Documents/Alex the analyst series/File sorter test/"
#path is the path to the folder containing unsorted files which we will sort with this code 
#the contense of a directory
file_name=os.listdir(path)
```

```python
folder_names = ['csv files', 'image files', 'text files'] # names for the folders to be created 

for loop in range(0,3):
    if not os.path.exists(path + folder_names[loop]): #if there isnt already files with the names in the folder_names list
         os.makedirs(path + folder_names[loop]) #create files with the names specified in the folder_names list
            
for file in file_name: 
    if ".csv" in file and not os.path.exists(path +"csv files/"+ file):   #if a csv is in the folder and not already in the new csv file we created 
        shutil.move(path + file, path+"csv files/"+file)                  #move the file into the new csv folder   
        
    elif ".png" in file and not os.path.exists(path +"image files/"+ file):   
        shutil.move(path + file, path+"image files/"+file) 
        
    elif ".txt" in file and not os.path.exists(path +"text files/"+ file):    
        shutil.move(path + file, path+"text files/"+file)
```
